```python
# we removing the columns which we dont want 

# We are making list for columns to drop and columns to keep

columns_to_keep = ['NAME', 'host id', 'host_identity_verified', 'host name',
       'neighbourhood group', 'neighbourhood', 'lat', 'long', 'country',
       'country code', 'instant_bookable', 'cancellation_policy', 'room type',
       'Construction year', 'price', 'service fee', 'minimum nights',
       'number of reviews', 'last review']

columns_to_drop = ['id','reviews per month',
       'review rate number', 'calculated host listings count',
       'availability 365', 'house_rules', 'license']
```

```python
len(columns_to_keep)
```

```python
len(columns_to_drop)
```

```python
#Filtering the data

data[columns_to_keep]  # data from columns to keep 
```

```python
data.shape
```

```python
# We didnt modified the original data set so we create new dataframe name "df" and value in it

df = data[columns_to_keep]
```

```python
df
```

```python
# 2nd method dropping columns :
# we have to delete "license" column but insted of deleting on columns we delete "columns_to_delete"

data.drop(columns = columns_to_drop, inplace = True)
```

```python
# Method 3 : check duplicated records

# data.duplicated().sum()   # show sum() of duplicates records
data.duplicated().value_counts()
```

```python
# Step 4 : Remove NaN value from columns

data.isna().sum()     # to check null or NaN value in columns it will return sum() NaN value also
```

```python
# Step 5 : Cleaning Induvidual columns

data
```

```python
# we have to to clean column name : "host_identity_verified"

data['host_identity_verified'].str.upper()
```

```python
# but if wrun our dataframe name :data the changes is not reflected for we need to save result in new variable

data['host_identity_verified'] = data['host_identity_verified'].str.upper()
```

```python
data
```

```python
# Step 5 : 

data['price']
```

```python
# here in "price" column we see $ sign but the data type of this column is string so we need to change it

data['price']= data['price'].str.replace("$", " ")   # replace $ with blank space we remove $ sign
```

```python
data['price']
```

```python
# now we want to remove comma (,)

data['price'].str.replace(",", "") 
```
